```python
# ============================================
#       STOCK PRICE DATA ANALYZER
#       Python Assignment - Google Colab
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 60)
print("          STOCK PRICE DATA ANALYZER")
print("=" * 60)


# --------------------------------------------
# 1. Create Sample Stock Price Dataset
# --------------------------------------------

data = {
    "Date": pd.date_range(start="2026-01-01", periods=15, freq="D"),

    "Open": [
        100, 102, 101, 105, 107,
        106, 109, 111, 110, 113,
        115, 114, 117, 119, 121
    ],

    "High": [
        104, 105, 106, 108, 110,
        109, 112, 114, 113, 116,
        118, 117, 120, 122, 124
    ],

    "Low": [
        98, 100, 99, 103, 105,
        103, 107, 109, 108, 111,
        113, 112, 115, 117, 119
    ],

    "Close": [
        102, 101, 105, 107, 106,
        109, 111, 110, 113, 115,
        114, 117, 119, 121, 123
    ],

    "Volume": [
        1200, 1500, 1300, 1800, 1600,
        1900, 2100, 2000, 2200, 2500,
        2300, 2700, 2900, 3100, 3300
    ]
}

df = pd.DataFrame(data)

print("\nStock Price Dataset:")
print(df)


# --------------------------------------------
# 2. Basic Information
# --------------------------------------------

print("\n" + "=" * 60)
print("              BASIC INFORMATION")
print("=" * 60)

print("Number of Records:", len(df))
print("Total Trading Volume:", df["Volume"].sum())


# --------------------------------------------
# 3. Statistical Analysis
# --------------------------------------------

print("\n" + "=" * 60)
print("             PRICE ANALYSIS")
print("=" * 60)

print(f"Average Opening Price : {df['Open'].mean():.2f}")
print(f"Average Closing Price : {df['Close'].mean():.2f}")

print(f"Highest Stock Price   : {df['High'].max():.2f}")
print(f"Lowest Stock Price    : {df['Low'].min():.2f}")

print(f"Closing Price Std Dev : {df['Close'].std():.2f}")


# --------------------------------------------
# 4. Daily Price Change
# --------------------------------------------

df["Daily Change"] = df["Close"].diff()

df["Daily Return (%)"] = df["Close"].pct_change() * 100

print("\n" + "=" * 60)
print("          DAILY PRICE CHANGES")
print("=" * 60)

print(df[[
    "Date",
    "Close",
    "Daily Change",
    "Daily Return (%)"
]].to_string(index=False))


# --------------------------------------------
# 5. Overall Return
# --------------------------------------------

initial_price = df["Close"].iloc[0]
final_price = df["Close"].iloc[-1]

overall_return = (
    (final_price - initial_price) / initial_price
) * 100

print("\n" + "=" * 60)
print("              RETURN ANALYSIS")
print("=" * 60)

print(f"Initial Closing Price : {initial_price:.2f}")
print(f"Final Closing Price   : {final_price:.2f}")
print(f"Overall Return        : {overall_return:.2f}%")


# --------------------------------------------
# 6. Maximum and Minimum Closing Price
# --------------------------------------------

max_close = df["Close"].max()
min_close = df["Close"].min()

max_date = df.loc[df["Close"].idxmax(), "Date"]
min_date = df.loc[df["Close"].idxmin(), "Date"]

print("\nHighest Closing Price:")
print(f"Price: {max_close:.2f}")
print(f"Date : {max_date.date()}")

print("\nLowest Closing Price:")
print(f"Price: {min_close:.2f}")
print(f"Date : {min_date.date()}")


# --------------------------------------------
# 7. Price Movement Chart
# --------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    df["Date"],
    df["Close"],
    marker="o",
    label="Closing Price"
)

plt.plot(
    df["Date"],
    df["Open"],
    marker="s",
    label="Opening Price"
)

plt.title("Stock Opening and Closing Prices")
plt.xlabel("Date")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# --------------------------------------------
# 8. Daily Return Chart
# --------------------------------------------

plt.figure(figsize=(10, 5))

plt.bar(
    df["Date"],
    df["Daily Return (%)"].fillna(0)
)

plt.title("Daily Stock Returns")
plt.xlabel("Date")
plt.ylabel("Return (%)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# --------------------------------------------
# 9. Trading Volume Chart
# --------------------------------------------

plt.figure(figsize=(10, 5))

plt.bar(
    df["Date"],
    df["Volume"]
)

plt.title("Trading Volume")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# --------------------------------------------
# 10. Final Conclusion
# --------------------------------------------

print("\n" + "=" * 60)
print("                  CONCLUSION")
print("=" * 60)

if overall_return > 0:
    print("The stock price increased during the analyzed period.")
elif overall_return < 0:
    print("The stock price decreased during the analyzed period.")
else:
    print("The stock price remained unchanged.")

print(f"Overall Return: {overall_return:.2f}%")
print(f"Average Closing Price: {df['Close'].mean():.2f}")
print(f"Price Volatility (Std Dev): {df['Close'].std():.2f}")

print("\nStock Price Data Analysis Completed Successfully!")
print("=" * 60)
```
